```python
import os
import pandas as pd
from xml.etree import ElementTree

data = []
reports_dir = '.' # Changed to current directory where XML files are located

# Ensure the directory exists before proceeding
if os.path.exists(reports_dir) and os.path.isdir(reports_dir):
    for filename in os.listdir(reports_dir):
        if filename.endswith('.xml'):  # Assuming reports are XML files
            full_path = os.path.join(reports_dir, filename)
            try:
                # Parse the XML file
                dom = ElementTree.parse(full_path)
                root = dom.getroot()

                # Extract text from AbstractText with Label='FINDINGS'
                findings_text = []
                for findings_elem in root.findall(".//Abstract/AbstractText[@Label='FINDINGS']"):
                    if findings_elem.text:
                        findings_text.append(findings_elem.text.strip())
                findings_text = ' '.join(findings_text)

                # Extract text from AbstractText with Label='IMPRESSION'
                impressions_text = []
                for impression_elem in root.findall(".//Abstract/AbstractText[@Label='IMPRESSION']"):
                    if impression_elem.text:
                        impressions_text.append(impression_elem.text.strip())
                impressions_text = ' '.join(impressions_text)

                data.append({
                    'filename': filename,
                    'FINDINGS': findings_text,
                    'IMPRESSIONS': impressions_text
                })
            except ElementTree.ParseError as e:
                print(f"Error parsing {filename}: {e}")
            except Exception as e:
                print(f"An unexpected error occurred with {filename}: {e}")
else:
    print(f"Error: Directory '{reports_dir}' not found or is not a directory. Please ensure NLMCXR_reports.tgz is uploaded and extracted.")

df = pd.DataFrame(data)

print(f"\nCollected {len(df)} XML file entries.")
display(df.head())
```

```python
from datasets import Dataset

df_dataset = Dataset.from_pandas(df)
```

```python
# tokenizer

def preprocess_function(batch):
  source = batch["FINDINGS"]
  target = batch["IMPRESSIONS"]
  source_ids = tokenizer(source, max_length=1024, truncation=True, padding="max_length", return_tensors="pt")
  target_ids = tokenizer(target, max_length=128, truncation=True, padding="max_length", return_tensors="pt")

  labels = target_ids["input_ids"].clone() # Create a copy
  labels[labels == tokenizer.pad_token_id] = -100

  return {"input_ids": source_ids["input_ids"], "attention_mask": source_ids["attention_mask"], "labels": labels}
```

```python
df_source = df_dataset.map(preprocess_function, batched=True)
```

```python
# Training Arguments

from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir='./content',
    num_train_epochs=2,
    per_device_train_batch_size=8,
    remove_unused_columns = True
)

split_dataset = df_dataset.train_test_split(test_size=0.2)
```

```python
trainer = Trainer(
    model = model,
    args = training_args,
    train_dataset = split_dataset["train"].map(preprocess_function, batched=True),
    eval_dataset = split_dataset["test"].map(preprocess_function, batched=True)
)
```

```python
trainer.train()
```

```python
eval_results = trainer.evaluate()
```

```python
tokenizer = AutoTokenizer.from_pretrained('content/model_directory')
model = AutoModelForSeq2SeqLM.from_pretrained('content/model_directory')

def summarize(FINDINGS):

  # tokenize the input findings
  inputs = tokenizer(FINDINGS, max_length=1024, truncation=True, return_tensors='pt')
  # generate the summary
  summary_ids = model.generate(inputs['input_ids'], max_length = 80, min_length=30, length_penalty=2.0, num_beams=4, early_stopping=True)
  # decode the summary
  summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
  return summary
```
